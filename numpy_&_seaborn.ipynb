```python
import numpy as np
```

```python
sales=np.array([45000,52000,48000,61000,75000,68000,72000,80000,85000,79000,92000,98000])
```

```python
print(sales)
```

```python
print("number of elements",sales.size)
print("dimensions",sales.ndim)
print("shape",sales.shape)
print("data type",sales.dtype)
```

```python
# calculate
print("total sales",sales.sum())
print("average sales",sales.mean())
print("minimum sales",sales.min())
print("maximun sales",sales.max())
print("standard deviation",sales.std())
```

```python
#highest sales
highest_month=np.argmax(sales)+1
highest_sales=sales.max
print("highest_month",highest_month)
print("highest_sales",highest_sales())
```

```python
#lowest sales
lowest_month=np.argmin(sales)+1
lowest_sales=sales.min
print("lowest_month",lowest_month)
print("lowest_sales",lowest_sales())

```

```python
sales_diff=sales.max()-sales.min()
print("The difference between the max and min",sales_diff)
```

```python
sales_increase=sales*1.10
print("orginal sales value",sales)
print("sales_increase by 10 %",sales_increase)
```

```python
month=np.where(sales>70000)[0]+1
print("high sales month",month)
```

```python
reshape_3x4=sales.reshape(3,4)
print("reshape",reshape_3x4)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Employee_ID":[101,102,103,104,105,106,107,108],
    "Name":["Arun","Priya","Rahul","Meena","Karthik","Divya","Vijay","Anu"],
    "Department":["IT","HR","IT","Finance","HR","IT","Finance","IT"],
    "Age":[25,28,32,29,35,26,40,31],
    "Salary":[45000,50000,65000,55000,60000,48000,75000,70000],
    "Experience":[2,3,7,5,10,2,15,8]
}

df = pd.DataFrame(data)
```

```python
sns.boxplot(y=df["Salary"])

plt.title("Box Plot of Salary")

plt.show()
```

```python
sns.histplot(df["Salary"], kde=True)

plt.title("Salary Distribution")

plt.show()
```

```python
sns.countplot(x="Department", data=df)

plt.title("Employees by Department")

plt.show()
```

```python
sns.scatterplot(x="Experience",
                y="Salary",
                hue="Department",
                data=df)

plt.title("Experience vs Salary")

plt.show()
```

```python
corr = df[["Age","Salary","Experience"]].corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Matrix")

plt.show()
```
